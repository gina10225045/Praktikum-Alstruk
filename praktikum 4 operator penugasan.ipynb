```python
# Poin awal
poin = 100
print("Poin awal :", poin)

# Tambahkan 25 poin menggunakan +=
poin += 25
print("Setelah ditambah 25 (+=)  :", poin)

# Kurangi 10 poin menggunakan -=
poin -= 10
print("Setelah dikurangi 10 (-=) :", poin)

# Kalikan poin dengan 2 menggunakan *=
poin *= 2
print("Setelah dikalikan 2 (*=)  :", poin)

# Bagi poin dengan 2 menggunakan /=
poin /= 2
print("Setelah dibagi 2 (/=)     :", poin)
```
