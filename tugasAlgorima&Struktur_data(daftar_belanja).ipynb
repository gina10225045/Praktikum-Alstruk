```python
daftar_belanja = []
def tambah_barang(nama_barang):
  daftar_belanja.append(nama_barang)
  print(f"'{nama_barang} ' berhasil ditambahkan.")

def lihat_daftar():
    print("\n=== DAFTAR BELANJAAN ===")
    if len (daftar_belanja) == 0:
      print("DAFTAR BELANJAAN KOSONG.")

    else:
        for nomor, barang in enumerate(daftar_belanja, start=1):
          print(f"{nomor}.{barang}")

tambah_barang("beras")
tambah_barang("minyak goreng")
tambah_barang("telur")
tambah_barang("mie instan")
tambah_barang("kopi")
tambah_barang("gula")
tambah_barang("susu")

lihat_daftar()
```
